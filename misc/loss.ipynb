This function defines the **L1 loss** (also called **Manhattan distance** or **mean absolute error**) commonly used in machine learning and statistics for regression problems.

Let's break it down:

---

### Mathematical Notation

$$
L_1(\hat{y}, y) = \sum_{i=0}^{m-1} \left| y^{(i)} - \hat{y}^{(i)} \right|
$$

* $y^{(i)}$ = the true value (ground truth) for the $i$-th example.
* $\hat{y}^{(i)}$ = the predicted value for the $i$-th example.
* $m$ = total number of examples.
* The vertical bars $| \cdot |$ indicate **absolute value**.

---

### What does it do?

* **Computes the sum of absolute differences** between each true value and its prediction across all examples.
* Measures how far off your predictions are from the actual values, regardless of direction (over/under-prediction).
* The lower the $L_1$ loss, the better your model's predictions match the actual data.

---

### Context in AI/ML

* Used as a **loss function** to optimize regression models.
* More robust to outliers than the L2 (squared error) loss, because it does not heavily penalize large errors.

---

### Example

Suppose you have:

* True values: $y = [2, 4, 6]$
* Predictions: $\hat{y} = [3, 4, 5]$

$$
L_1 = |2-3| + |4-4| + |6-5| = 1 + 0 + 1 = 2
$$

---

### Summary

* **L1 loss** sums the absolute prediction errors over all samples.
* Used for **regression** and **robust fitting** in machine learning.


```python
import numpy as np

# True values and predictions
y_true = np.array([2, 4, 6])
y_pred = np.array([3, 4, 5])

# L1 Loss calculation (sum of absolute errors)
l1_loss = np.sum(np.abs(y_true - y_pred))

print("L1 Loss:", l1_loss)
```

Implement the numpy vectorized version of the L2 loss. There are several way of implementing the L2 loss but you may find the function np.dot() useful. As a reminder, if $x = [x_1, x_2, ..., x_n]$, then `np.dot(x,x)` = $\sum_{j=1}^n x_j^{2}$. 

- L2 loss is defined as $$\begin{align*} & L_2(\hat{y},y) = \sum_{i=0}^{m-1}(y^{(i)} - \hat{y}^{(i)})^2 \end{align*}\tag{7}$$

```python
import numpy as np

# True values and predictions
y_true = np.array([2, 4, 6])
y_pred = np.array([3, 4, 5])

# L2 Loss calculation (sum of squared errors) using np.dot (vectorized)
l2_loss = np.dot(y_true - y_pred, y_true - y_pred)

print("L2 Loss:", l2_loss)
```
